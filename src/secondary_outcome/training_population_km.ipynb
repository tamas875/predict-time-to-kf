# Save the population Kaplan-Meier model

```python
import pandas as pd
import numpy as np
import os
from dotenv import load_dotenv

load_dotenv()

train_data_path = os.getenv('TRAINING_DATA_SECONDARY_OUTCOME')
train = pd.read_csv(train_data_path)
train = train.drop(columns=['Studio', 'ESRD', 'Creat1', 'esrd_upper', 'esrd_lower'])
train.shape
```

```python
train.columns
```

```python
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter

# Visualize the survival distribution
kmf = KaplanMeierFitter()
kmf.fit(durations=train['fu_morte_1'], event_observed=train['Death_preHD'])
kmf.plot_survival_function()
plt.show()
```

```python
import joblib
joblib.dump(kmf, 'kmf.joblib')
```
